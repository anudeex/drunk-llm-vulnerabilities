```python
with open("./data/TFLN-drunk-texts.txt", "r") as f:
    drunk_texts = [line.rstrip("\n") for line in f]
with open("./data/TFLN-non-drunk-texts.txt", "r") as f:
    non_drunk_texts = [line.rstrip("\n") for line in f]
print("Drunk texts:", len(drunk_texts), " Non Drunk texts:", len(non_drunk_texts))
X = drunk_texts + non_drunk_texts
y = [1]*len(drunk_texts) + [0]*len(non_drunk_texts)
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.linear_model import LogisticRegression

# X = list of texts, y = labels (0/1)
sbert_model = SentenceTransformer('all-MiniLM-L6-v2')
X_sbert = sbert_model.encode(X)
```

```python
clf_sbert = LogisticRegression()
```

```python
from sklearn.model_selection import cross_val_predict, StratifiedKFold, cross_val_score
from sklearn.metrics import classification_report

cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

y_pred = cross_val_predict(
    clf_sbert,
    X_sbert,
    y,
    cv=cv
)

# Classification report
print("10-Fold Cross-Validation Results")
print(classification_report(y, y_pred))

scores = cross_val_score(
    clf_sbert,
    X_sbert,
    y,
    cv=cv,
    scoring="precision"
)
print("Mean precision:", scores.mean())

scores = cross_val_score(
    clf_sbert,
    X_sbert,
    y,
    cv=cv,
    scoring="recall"
)
print("Mean recall:", scores.mean())

scores = cross_val_score(
    clf_sbert,
    X_sbert,
    y,
    cv=cv,
    scoring="f1"
)
print("Mean f1:", scores.mean())
```

```python
clf_sbert.fit(X_sbert, y)
```

```python
import pickle

# Save model
with open('logreg_sbert_model.pkl', 'wb') as f:
    pickle.dump(clf_sbert, f)
```
